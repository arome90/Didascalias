```python
import pandas as pd
import matplotlib.pyplot as plt


def graficar_columna(df, columna):
    nombres = df['Name'].unique()
    
    plt.figure(figsize=(10,5))
    for nombre in nombres:
        df_nombre = df[df['Name'] == nombre]
        plt.plot(df_nombre['Actions'], df_nombre[columna], label=nombre) 
    plt.xlabel('Time')
    plt.ylabel(columna)
    plt.title(f'Gráfica de {columna} por Nombre')
    plt.legend()
    plt.grid(True)
    plt.show()


def graficar_columna(df,name, columna):

    df_nombre = df[df['Name'] == nombre]
    plt.plot(df_nombre['Actions'], df_nombre[columna], label=nombre) 
    plt.xlabel('Time')
    plt.title(f'Gráfica de {columna} de {name}')
    plt.legend()
    plt.grid(True)
    plt.show()

df = pd.read_csv('2025-03-15_23-49-15_students_datas.csv')

c_excluir = ['Time','Name','Attention']
nombres = df['Name'].unique()

for nombre in nombres:
    df_nombre = df[df['Name'] == nombre]  
    plt.figure(figsize=(10,5))
    
    for columna in df.columns:
        if columna in c_excluir:
            continue
        plt.plot(df_nombre['Time'], df_nombre[columna], label=columna)
    
    plt.xlabel('Time')
    plt.ylabel('Valores')
    plt.title(f'Gráfica de {nombre}')
    plt.legend()
    plt.grid(True)
    plt.show()

    graficar_columna(df, nombre, 'Attention')
```
